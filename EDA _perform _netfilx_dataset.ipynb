# Univariate Analysis

```python
# using categorical
# countploat 
# type of movie me alwalye popele see the movie 
# because t.v show are very long (like seasone and epsoide)
sns.countplot(df["Type"])
plt.title("Movie vs TV show ")
```

```python
# top 10 Director	
sns.barplot(df["Director"].head(10),color="yellow")
plt.title("TOP 10 Director")


```

```python
# top 10 acctors
sns.barplot(df["Cast"].head(10))
plt.title("TOP 10 Actors")
```

```python
# top 10 country 
sns.barplot(df["Country"].head(10))
plt.title("TOP 10 country")
```

```python
# top ten date 
sns.countplot(df["Date_added"].head(10))
plt.title("TOP 10 date_added")
```

# movies release_years 

```python
# movies release_years 
plt.figure(figsize=(15,4))
sns.scatterplot(df["Release_year"].value_counts())
plt.xticks(rotation=90) 
plt.title('RELEASE YEARS')
plt.show()
# higest movies relase_years ( 2000 > 2020 linear groth)
```

```python
# pdf
plt.figure(figsize=(10,5))

sns.kdeplot(df["Release_year"],fill=True,color="red")

plt.title("Probability Density Function (PDF)")
plt.xlabel("Release Year")
plt.ylabel("Density")

plt.show()
```
